# Student Academic Performance Analysis System
**Technologies:** Python, NumPy, Pandas  
**Level:** Beginner

This notebook analyzes student marks, calculates totals and percentages, assigns grades, finds the top and lowest performers, shows subject-wise averages, and saves a CSV report.

## 1. Import Libraries

```python
import numpy as np
import pandas as pd
```

## 2. Create the Student Dataset

```python
data = {
    "Roll_No": [1, 2, 3, 4, 5],
    "Name": ["Aman", "Riya", "Rahul", "Priya", "Sneha"],
    "Python": [85, 90, 65, 78, 95],
    "Maths": [80, 95, 70, 85, 92],
    "Physics": [75, 88, 60, 80, 96],
    "English": [90, 92, 75, 85, 98]
}

df = pd.DataFrame(data)

subjects = ["Python", "Maths", "Physics", "English"]
maximum_marks = 100
number_of_subjects = len(subjects)

df
```

## 3. Calculate Total, Average, and Percentage

```python
df["Total"] = np.sum(df[subjects].to_numpy(), axis=1)
df["Average"] = np.mean(df[subjects].to_numpy(), axis=1)

maximum_total = maximum_marks * number_of_subjects
df["Percentage"] = (df["Total"] / maximum_total) * 100

df
```

## 4. Assign Grades

```python
def calculate_grade(percentage):
    """Return a grade based on the student's percentage."""
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 40:
        return "D"
    else:
        return "Fail"


df["Grade"] = df["Percentage"].apply(calculate_grade)
df
```

## 5. Display the Student Result

```python
print("\nSTUDENT ACADEMIC PERFORMANCE ANALYSIS")
print("=" * 55)
print("\nSTUDENT DATA:")
print(df.to_string(index=False))

print("\nTOP PERFORMER:")
top_student = df.loc[df["Total"].idxmax()]
print("Name       :", top_student["Name"])
print("Total      :", top_student["Total"], "/", maximum_total)
print("Percentage :", f'{top_student["Percentage"]:.2f}%')
print("Grade      :", top_student["Grade"])

print("\nLOWEST PERFORMER:")
lowest_student = df.loc[df["Total"].idxmin()]
print("Name       :", lowest_student["Name"])
print("Total      :", lowest_student["Total"], "/", maximum_total)
print("Percentage :", f'{lowest_student["Percentage"]:.2f}%')
print("Grade      :", lowest_student["Grade"])
```

## 6. Subject-wise Analysis

```python
subject_average = df[subjects].mean()

print("\nSUBJECT-WISE AVERAGE MARKS:")
print(subject_average.round(2).to_string())

print("\nHIGHEST AVERAGE SUBJECT:", subject_average.idxmax())
print("LOWEST AVERAGE SUBJECT :", subject_average.idxmin())
```

## 7. Grade Count

```python
print("\nNUMBER OF STUDENTS IN EACH GRADE:")
print(df["Grade"].value_counts().to_string())
```

## 8. Save the Final Report as a CSV File

```python
df.to_csv("student_performance_report.csv", index=False)

print("\n" + "=" * 55)
print("Project completed successfully!")
print("CSV report saved as: student_performance_report.csv")
```
